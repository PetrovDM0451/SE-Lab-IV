```python
BP = {
    'PM25': [(0,12,0,50),(12.1,35.4,51,100),(35.5,55.4,101,150),
             (55.5,150.4,151,200),(150.5,250.4,201,300),(250.5,500.4,301,500)],
    'PM10': [(0,54,0,50),(55,154,51,100),(155,254,101,150),
             (255,354,151,200),(355,424,201,300),(425,604,301,500)],
    'CO':   [(0,4.4,0,50),(4.5,9.4,51,100),(9.5,12.4,101,150),
             (12.5,15.4,151,200),(15.5,30.4,201,300),(30.5,50.4,301,500)],
    'NO2':  [(0,53,0,50),(54,100,51,100),(101,360,101,150),
             (361,649,151,200),(650,1249,201,300),(1250,2049,301,500)],
}

def sub_aqi(c, table):
    if np.isnan(c): return np.nan
    for (Clo, Chi, Ilo, Ihi) in table:
        if Clo <= c <= Chi:
            return (Ihi - Ilo) / (Chi - Clo) * (c - Clo) + Ilo
    return 500.0   # above highest breakpoint

df_clean['CO_ppm']  = df_clean['CO']  / 1145
df_clean['NO2_ppb'] = df_clean['NO2'] / 1.88

df_clean['AQI_PM25'] = df_clean['PM2_5'].apply(lambda x: sub_aqi(x, BP['PM25']))
df_clean['AQI_PM10'] = df_clean['PM10'].apply(lambda x: sub_aqi(x, BP['PM10']))
df_clean['AQI_CO']   = df_clean['CO_ppm'].apply(lambda x: sub_aqi(x, BP['CO']))
df_clean['AQI_NO2']  = df_clean['NO2_ppb'].apply(lambda x: sub_aqi(x, BP['NO2']))

AQI_COLS = ['AQI_PM25','AQI_PM10','AQI_CO','AQI_NO2']
df_clean['AQI'] = df_clean[AQI_COLS].max(axis=1).round(1)

df_clean[['Date','AQI_PM25','AQI_PM10','AQI_CO','AQI_NO2','AQI']].head(8)
```

```python
def classify_aqi(aqi):
    if aqi <= 50:
        return 'Good'
    elif aqi <= 100:
        return 'Moderate'
    elif aqi <= 150:
        return 'Unhealthy for Sensitive Groups'
    elif aqi <= 200:
        return 'Unhealthy'
    elif aqi <= 300:
        return 'Very Unhealthy'
    else:
        return 'Hazardous'

df_clean['AQI_Category'] = df_clean['AQI'].apply(classify_aqi)

summary_table = df_clean['AQI_Category'].value_counts().reset_index()

summary_table.columns = ['AQI_Category', 'Hours']

summary_table['%'] = (
    summary_table['Hours'] / len(df_clean) * 100
).round(1)

category_order = [
    'Good',
    'Moderate',
    'Unhealthy for Sensitive Groups',
    'Unhealthy',
    'Very Unhealthy',
    'Hazardous'
]

summary_table['AQI_Category'] = pd.Categorical(
    summary_table['AQI_Category'],
    categories=category_order,
    ordered=True
)

summary_table = summary_table.sort_values('AQI_Category')

print(summary_table.to_string(index=False))
```
